## 设计思路：
1. 所有标准工作时间以外的时长均算作加班时长
2. 每天的标准工时是8小时，超过8小时以外的时间算调休时长
3. 工作日中午12:00-13:30为午休时长，在计算时长时，午休时长不参与计算
4. 以凌晨三点为分界点，凌晨三点之前的算作上一天的上班时长中
5. 需要获取当年的国家节假日安排，所有节假日时间的上班打卡记录，均算作加班时长，没有午休时长规则扣除
6. 当前的excel考勤报表中会将周日设置为工作日，实际计算时，在遇到是周日且非国家规定的节假日调休的工作日，这种情况下考勤统计按照节假日的规则，也就是没有午休时长，实际的打卡时长就是加班时长
7. 系统导出的考勤报表放在export_report中，计算之后的excel文件新建一个文件夹以姓名-年月日的形式命名存放
8. 需要做一个m系列芯片的dmg安装包，mac可以安装运行的。由于本机没有获得签名许可，因此同时需要有一个可以运行之后生成dmg的脚本，以便别的同事拿到手可以快速运行生成所需dmg。
9. 需要做一个x86版本windows可以直接双击运行的exe文件，以便使用windows笔记本的同事可以直接使用。
10. 安装包的需求最后等全部验证完毕之后确认了再做
11. 如果某天只有上下班打卡记录缺少其中一个则跳过改天的计算
12. 存在多个打卡记录的，以最早上班打卡时间和最晚下班打卡时间为准
13. 在打卡详情中，只要是上班打卡记录或者下班打卡记录中存在打卡地点为：弘快科技 | 上海市普陀区云岭东路586号 的，则当天的工作时长减半小时，这个是公司给我们员工的异地办公福利。但是节假日没有该福利，因为节假日加班午休时长已经给算作加班时长了。
14. 统计结果中的数据需要具备一目了然的特性，同时又可以看到每天的详细情况；表现形式你思考一下，建议可以是那种用户选择源文件之后，计算完毕之后以一个客户端页面展示的那种，页面的话我倾向于做成更现代的风格
